# Fila

````python

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
    
    # Inclusão de elementos na fila
    def enqueue(self, value):
        new_node = Node(value)

        if self.front == None:
            self.front = self.rear = new_node
            return
        self.rear.next = new_node
        self.rear = new_node

    # Remoção de elemnto da fila
    def dequeue(self):
        if self.front == None:
            raise Exception("Queue is empty")
        temp = self.front
        self.front = self.front.next
        if self.front == None:
            self.rear = None
        return temp.value

    # Acessa o primeiro elemento da fila
    def peek(self):
        if self.front == None:
            raise Exception("Queue is empty")
        return self.front.value
    
    # Verifica se a fila está vazia
    def is_empty(self):
        return self.front == None

    # Exibe a fila
    def display_queue(self):
        current = self.front
        while current != None:
            print(current.value, end=" ")
            current = current.next

# Teste da classe Queue
queue = Queue()

# Enfileirar elementos iniciais
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)

# Enfileirar mais elementos
queue.enqueue(4)
queue.enqueue(5)
queue.enqueue(6)

# Exibir a fila após enfileirar
print("Fila após enfileirar 4, 5 e 6:")
queue.display_queue()
print()

# Desenfileirar elementos
queue.dequeue()
queue.dequeue()

# Exibir a fila após desenfileirar
print("Fila após desenfileirar duas vezes:")
queue.display_queue()
print()

# Visualizar o primeiro elemento
print("Primeiro elemento da fila (peek):", queue.peek())

# Verificar se a fila está vazia
if queue.is_empty():
    print("Fila está vazia")
else:
    print("Fila não está vazia")
```

````
